# Creating a Function

```python
# Creating a fucntion to convert the surival time months into years and making a new column with it
def calculate_survival_years(row):
    months = row['Overall Survival (Months)']
    # Survival in years
    if months == 0:
        return 0
    return months / 12

#.apply() to create the new column by using the function
df_clinical['Overall Survival (Years)'] = df_clinical.apply(calculate_survival_years, axis=1)
```

# Uploading Mutation Dataframe

```python
# This dataframe gives more genetic information on the samples collected from the patients
url = "https://github.com/vanessadiaz04/python_capstone_pancreatic_cancer/raw/refs/heads/main/data_mutations.csv"
df_hugo = pd.read_csv(url)
```

```python
df_hugo
```

```python
#'Missense_Mutation'= A single nucleotide change that results in a different amino acid being incorporated into the protein
#'Nonsense_Mutation'= A single nucleotide change that introduces a premature stop codon, leading to a shortened (truncated) and usually nonfunctional protein
#'Frame_Shift_Del'= A deletion of nucleotides (not in a multiple of 3) that shifts the reading frame of the gene, changing all subsequent amino acids.
#'Frame_Shift_Ins'= An insertion of nucleotides (not in a multiple of 3) that shifts the reading frame.
#'In_Frame_Del'= A deletion of nucleotides in a multiple of 3, resulting in the loss of amino acids but preserving the overall reading frame.
#'In_Frame_Ins'=  An insertion of nucleotides in a multiple of 3, adding amino acids without shifting the reading frame.
#'Translation_Start_Site'= A mutation at the start codon (usually ATG) that can prevent or alter the start of protein translation.
#"5'Flank"=  Mutations in the regions flanking the gene, typically upstream (5')
#"3'UTR" and "5'UTR" = Mutations in the 3' or 5' UnTranslated Regions.
  #These are transcribed into mRNA but not translated into protein, and they regulate mRNA stability and translation efficiency.
#'Splice_Region'= A mutation in the region adjacent to the splice site but not directly on the splice junction
#'Intron'= A mutation occurring within the non-coding intronic region between exons
#'Nonstop_Mutation'= A mutation that removes the normal stop codon, causing protein translation to continue into the 3' UTR, creating an abnormally long protein
#"3'Flank"= Mutations in the regions flanking the gene, typically downstream (3')

df_hugo["Variant_Classification"].unique()
```

```python
#SNP: Single Nucleotide Polymorphism (single base change).
#DEL: Deletion (loss of bases).
#INS: Insertion (gain of bases).
#DNP: Dinucleotide Polymorphism (two consecutive base changes).
#ONP: Oligonucleotide Polymorphism (multiple consecutive base changes).
df_hugo["Variant_Type"].unique()
```

```python
#t_ref_count stands for Tumor Reference Count.
# It represents the number of sequencing reads in a tumor sample that support the reference allele (the original, non-mutated base) at a specific genomic position
df_hugo["t_ref_count"].unique()
```

```python
#n_ref_count = normal reference count.
#It represents the number of sequencing reads from a normal (non-tumor/germline) sample that support the reference allele (the standard base at that specific genomic location).
exists = (df_hugo['n_ref_count'] == 'NaN').any()
exists
```

```python
df_hugo = df_hugo.drop(columns=['Center',
                                'NCBI_Build',
                                'Strand',
                                'dbSNP_Val_Status',
                                'Matched_Norm_Sample_Barcode',
                                'Match_Norm_Seq_Allele1',
                                'Match_Norm_Seq_Allele2',
                                'Tumor_Validation_Allele1',
                                'Tumor_Validation_Allele2',
                                'Match_Norm_Validation_Allele1',
                                'Match_Norm_Validation_Allele2',
                                'Verification_Status',
                                'Validation_Status',
                                'Mutation_Status',
                                'Sequencing_Phase',
                                'Sequence_Source',
                                'Validation_Method',
                                'Score',
                                'BAM_File',
                                'Sequencer',
                                'Exon_Number',
                                'COMMENTS',
                                'PATH_SCORE'])
```

```python
# function to calculate the Variant Allele Frequency (VAF).
#The output measures the proportion of sequencing reads supporting a specific mutation compared to the total read depth at that genomic locus.
# It represents the fraction of tumor DNA (allelic burden) within a sample.
# VAF interprets tumor heterogeneity, purity, and mosaicism, with higher values indicating germline or pure tumor variants
# lower values suggesting subclonal population
def calculate_VAF(row):
    alt = row['t_alt_count']
    ref = row['t_ref_count']
    # Variant Allele Frequency (VAF)
    if (alt + ref) == 0:
        return 0
    return alt / (alt + ref)

#.apply() to create the new column by using the function
df_hugo['VAF'] = df_hugo.apply(calculate_VAF, axis=1)
```
