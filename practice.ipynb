# Linear Regression

```python
df_linear = df_clinical.dropna(subset=['Overall Survival (Months)'])
```

```python
df_linear
```

```python
df_linear = df_linear[df_linear['Neoadjuvant Treatment'] != 'Other']
df_linear = df_linear[df_linear['Adjuvant Treatment'] != 'Other']
```

```python
import statsmodels.api as sm
import numpy as np
```

```python
y = df_linear['Overall Survival (Months)'].to_numpy()
X = sm.add_constant(df_linear['TMB (nonsynonymous)'].to_numpy())
model = sm.OLS(y, X).fit()

X_predict = np.array([1, 100_000])
predictions = model.predict(X_predict)
predictions
```

```python
model.summary()
```

```python
sns.scatterplot(data=df_linear, x='Fraction Genome Altered', y='Overall Survival (Months)', hue='KRAS Somatic Variant')
```

```python
n = 10
trials = 90

sample_means = [df_linear['Overall Survival (Months)'].sample(n, replace=True).mean() for _ in range(trials)]

means_series = pd.Series(sample_means)
sns.histplot(means_series, kde=True)
plt.title(f"Distribution of Survival Time (months) (n={n})")
plt.show()
```

##Logistic Regression

```python
df_cleaned = df_clinical
```

```python
df_cleaned = df_cleaned[['KRAS Allele Selection State','KRAS Mutant Copies Gained','Overall Survival','KRAS Somatic Variant','Tumor Location','Fraction Genome Altered', 'Age at Diagnosis', 'TMB (nonsynonymous)']]
```

```python
df_cleaned.replace('1:DECEASED', 'DECEASED', inplace=True)
```

```python
df_cleaned.replace('0:LIVING', 'LIVING', inplace=True)
```

```python
df_cleaned['KRAS Mutant Copies Gained'].unique()
```

```python
df_cleaned.replace('5+', '5', inplace=True)
```

```python
df_cleaned = df_cleaned[df_cleaned['KRAS Somatic Variant'] != 'Other']
```

```python
df_cleaned["KRAS Somatic Variant"].unique()
```

```python
KRAS_Somatic_Variant = pd.get_dummies(df_cleaned["KRAS Somatic Variant"], drop_first=True, dtype=int)
```

```python
Tumor_Location = pd.get_dummies(df_cleaned["Tumor Location"], drop_first=True, dtype=int)
```

```python
df_cleaned["KRAS Allele Selection State"].unique()
```

```python
KRAS_Allele_Selection_State = pd.get_dummies(df_cleaned["KRAS Allele Selection State"], drop_first=True, dtype=int)
```

```python
KRAS_Allele_Selection_State
```

```python
df_cleaned
```

```python
Overall_Survival = pd.get_dummies(df_cleaned["Overall Survival"], drop_first=True, dtype=int)
```

```python
Overall_Survival.rename(columns={'LIVING': 'Overall_Survival'}, inplace=True)
```

```python
Overall_Survival
```

```python
df_cleaned.drop(columns=["KRAS Allele Selection State", "KRAS Somatic Variant", "Tumor Location", "Overall Survival"], inplace=True)
```

```python
df_cleaned
```

```python
df_mutations = pd.concat([KRAS_Allele_Selection_State, KRAS_Somatic_Variant, Tumor_Location, Overall_Survival], axis=1)
```

```python
df_mutations
```

```python
df = pd.concat([df_cleaned, df_mutations], axis=1)
```

```python
df
```

```python
train = df
```

```python
train
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    train.drop(columns=["Overall_Survival"]),
    train["Overall_Survival"],
    test_size=0.10,
    random_state=101
    )
```

```python
from sklearn.linear_model import LogisticRegression

logmodel = LogisticRegression()
```

```python
logmodel.fit(X_train, y_train)
```

```python
predictions = logmodel.predict(X_test)
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, predictions))
```

```python
X_train.columns
```

```python
import numpy as np

coef_df = pd.DataFrame({
    'feature': train.drop("Overall_Survival", axis=1).columns,
    'coefficient': logmodel.coef_[0],
    'abs_coefficient': np.abs(logmodel.coef_[0])
}).sort_values('abs_coefficient', ascending=False)
```
